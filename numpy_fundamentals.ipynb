### Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
a1
```

```python
a1
```

```python
# Scaler Operations
# Arithmetic
a1 ** 2
```

```python
# relational
a2 > 15
```

```python
# Vector Operations
# arithmetic
a1 * a2
```

### Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod
# 0 -> col and 1 -> row
np.min(a1, axis=1)
```

```python
# mean/median/std/var
np.std(a1, axis=1)
```

```python
# trigonomoetrix function
np.sin(a1)
```

```python
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

np.dot(a2, a3)
```

```python
# log and exponets
np.exp(a1)
```

```python
# round/floor/ceil/trunc

np.trunc(np.random.random((2,3))*100)
```
